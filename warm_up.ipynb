```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.normal(0,1,1000)
    
n_bins = 100
fig, ax = plt.subplots()
ax.set_title('normal')
hist = ax.hist(x, bins=n_bins)
plt.show()


y=range(0,1000)
plt.plot(y,x,linestyle = 'dotted')

# Add title and axis labels
plt.title('Line Plot Example')
plt.xlabel('index')
plt.ylabel('Value')

# Display the plot
plt.show()
```

add trend line

```python
from sklearn.linear_model import LinearRegression
```

```python
trend_line=np.arange(1000)*.01
y_trend=x+trend_line

X, Y = trend_line.reshape(-1,1), y_trend.reshape(-1,1)

plt.scatter(trend_line,y_trend)
plt.plot( X, LinearRegression().fit(X, Y).predict(X),color='red' )

# Add title and axis labels
plt.title('Line Plot Example')
plt.xlabel('index')
plt.ylabel('Value')

# Display the plot
plt.show()
```

3-cumsum of noise

```python
new_x=[]
new_x.append(x[0])
for i in range(1,len(x)):
    new_x.append(new_x[i-1]+x[i])
    
y=range(0,1000)
plt.plot(y,new_x,linestyle = 'dotted')

# Add title and axis labels
plt.title('Line Plot Example')
plt.xlabel('index')
plt.ylabel('Value')

# Display the plot
plt.show()
   

    
```

4- multivariate normal

```python
mean=np.array([0,0])
cov=np.array([[1,-.5],[-.5,2]])
sample=np.random.multivariate_normal(mean,cov,1000)

plt.Figure(figsize=(8,6))
plt.scatter(sample[:,0],sample[:,1],alpha=.5)
plt.show()
```

5-mean and covarience form multivariate_normal

```python

mean_x=sample[:,0].sum()/1000
mean_y=sample[:,1].sum()/1000

var_x= ((sample[:,0]-mean_x)**2).sum()/999
var_y= ((sample[:,1]-mean_y)**2).sum()/999
cov_xy=((sample[:,1]-mean_y)*(sample[:,0]-mean_x)).sum()/999
```

Price simulation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T=1000
p0=10
mu=.001   #drift
```

```python
last_p=np.log(p0)
log_returns=np.zeros(T)
prices=np.zeros(T)
for t in range(T):
    r=.01*np.random.randn()
    p=last_p+mu+r
    log_returns[t]=r+mu
    prices[t]=np.exp(p)
    last_p=p
```

```python
plt.figure(figsize=(20,8))
plt.plot(prices)
plt.title("Simulated Prices")
```
